# Dice simulation

Python version of `dice.m`. Roll a fair six-sided die `no_trials` times and track how often each face
appears. Each face should approach 1/6 ≈ 16.7% of rolls.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(1)   # fixed seed = same results each run
no_trials = 500

trials = rng.integers(1, 7, size=no_trials)          # integers 1..6
n = np.arange(1, no_trials + 1)

cumulative_count = np.column_stack([np.cumsum(trials == face) for face in range(1, 7)])
cumulative_per = cumulative_count / n[:, None] * 100

for face in range(1, 7):
    print(f"{face}: {cumulative_count[-1, face - 1]:4d} rolls ({cumulative_per[-1, face - 1]:.1f}%)")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
for face in range(1, 7):
    axes[0].plot(n, cumulative_count[:, face - 1], label=f"No. {face}s")
    axes[1].plot(n, cumulative_per[:, face - 1], label=f"No. {face}s")
axes[0].set(xlabel="Trials", ylabel="No. outcomes", title="No. outcomes per trial")
axes[1].axhline(100 / 6, color="grey", linestyle="--")
axes[1].set(xlabel="Trials", ylabel="Percentage", title="Percentage of outcomes per trial")
axes[0].legend(); axes[1].legend()
plt.tight_layout()
plt.show()
```
