# Exercise 3

a) Solve the following system of equations by hand using elimination or substitution.

$
\begin{align}
3x+y&=11 \\
x+4y&=11
\end{align}
$

b) Solve using matrix form

c) Use Python to solve this system.


```python
import numpy as np

A = np.array([[3, 1], [1, 4]])
b = np.array([11, 11])

x = np.linalg.solve(A, b)
print(x)   # [3. 2.]
```

# Exercise 4

Produce a plot of your solution using Python

Amend the code below.


```python
import matplotlib.pyplot as plt

# Here it is in python!

# This is a square matrix with 2 rows and 2 columns (2,2).
A = np.array([[3, 1],
              [1, 4]])

b = np.array([11, 11])

# We can solve a system of linear equations using the np.linalg.det() function.
x = np.dot(np.linalg.inv(A), b)

X = np.linspace(x[0] - 5, x[0] + 5, num=100) # It solves for the intersection point: x becomes [3, 2].
Y1 = (b[0] - A[0,0]*X)/A[0,1] # rearranges the first equation into y = 11 - 3x.
Y2 = (b[1] - A[1,0]*X)/A[1,1] # rearranges the second equation into y=(11-x)/4

plt.plot(X,Y1)
plt.plot(X,Y2)
plt.plot(x[0], x[1], "ko")
plt.xlabel("$x$")
plt.ylabel("$y$")

plt.show()
```

# Exercise 7

Solve the following system using Python.

$
\begin{align}
2x+3y-2z&=3 \\
x-2y+4z&=0 \\
2x-3y+z&=1
\end{align}
$

You will need to work out $\mathbf{A}$ and $\mathbf{b}$.

$
\begin{align}
  \mathbf{A}\mathbf{x} &= \mathbf{b} \\
  \begin{pmatrix}
  * & * & * \\
  * & * & * \\
  * & * & *
  \end{pmatrix}
  \begin{pmatrix}
  x \\
  y \\
  z \\
  \end{pmatrix}
  &=
  \begin{pmatrix}
  * \\
  * \\
  *
  \end{pmatrix}
\end{align}
$


```python
import numpy as np

A = np.array([[2, 3, -2],
              [1, -2, 4],
              [2, -3, 1]])
b = np.array([3, 0, 1])

x = np.linalg.solve(A, b)
print(x)                  # [ 0.97435897  0.28205128 -0.1025641 ]
print(np.linalg.det(A))   # about 39
```

# Exercise 8

Visualise your solution by amending the following code.

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

A = np.array([[2, 3, -2],
              [1, -2, 4],
              [2, -3, 1]])

b = np.array([3, 0, 1])

# We can solve a system of linear equations using the np.linalg.det() function.
x = np.dot(np.linalg.inv(A), b)

X = np.linspace(-10, 10, num=20)
Y = np.linspace(-10, 10, num=20)
xv, yv = np.meshgrid(X, Y)

z1 = (b[0] - A[0,0]*xv - A[0,1]*yv)/A[0,2];
z2 = (b[1] - A[1,0]*xv - A[1,1]*yv)/A[1,2];
z3 = (b[2] - A[2,0]*xv - A[2,1]*yv)/A[2,2];


fig, ax = plt.subplots(subplot_kw={"projection": "3d"})

# https://matplotlib.org/stable/api/_as_gen/mpl_toolkits.mplot3d.axes3d.Axes3D.view_init.html
ax.view_init(0, 0) # use this to manually rotate your plot
ax.plot_surface(xv,yv,z1)
ax.plot_surface(xv,yv,z2)
ax.plot_surface(xv,yv,z3)

plt.show()
```

I have also provided an interactive version of the plot.

```python
from ipywidgets import interact, fixed
# Importing mplot3d toolset
from mpl_toolkits.mplot3d import Axes3D



@interact(elev=(-90, 90), azim=(-180, 180))
def plot_3D(elev=30, azim=30):
  A = np.array([[2, 3, -2],
              [1, -2, 4],
              [2, -3, 1]])

  b = np.array([3, 0, 1])
  # We can solve a system of linear equations using the np.linalg.det() function.
  x = np.dot(np.linalg.inv(A), b)

  # Data for three-dimensional scattered points
  X = np.linspace(-10, 10, num=20)
  Y = np.linspace(-10, 10, num=20)
  xv, yv = np.meshgrid(X, Y)

  z1 = (b[0] - A[0,0]*xv - A[0,1]*yv)/A[0,2];
  z2 = (b[1] - A[1,0]*xv - A[1,1]*yv)/A[1,2];
  z3 = (b[2] - A[2,0]*xv - A[2,1]*yv)/A[2,2];


  fig, ax = plt.subplots(subplot_kw={"projection": "3d"})

  # https://matplotlib.org/stable/api/_as_gen/mpl_toolkits.mplot3d.axes3d.Axes3D.view_init.html
  ax.view_init(elev, azim) # use this to manually rotate your plot
  ax.scatter(x[0], x[1], x[2], c='k', s=20)
  ax.plot_surface(xv,yv,z1, alpha=0.3)
  ax.plot_surface(xv,yv,z2, alpha=0.3)
  ax.plot_surface(xv,yv,z3, alpha=0.3)


  plt.show()
```
