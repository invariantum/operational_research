# Tutorial 2, Q5c: product X takes 4.5 hours

**The problem.** Q5b, but product X now takes 4.5 hours instead of 5:

maximise $12x + 5y + 8z$ subject to $4.5x + 2y + 3z \le 100$, $z \le 20$, $x, y, z \ge 0$.

**How to run it.** *Run → Run All Cells*, or press Shift + Enter in each cell while you read. Nothing to install: SciPy is built into the browser's Python (the first import takes a few seconds).

**Expected result.** `x[1] = 22.22`, `x[2] = 0.00`, `x[3] = 0.00`, `fval = -266.67`: 22.22 of X, profit
£266.67. `linprog` minimises, so the printed `fval` is **minus** the maximum profit (here −266.67 means
£266.67).

Script name says 4c: the question was Q4 on an earlier sheet.

*Original script: `5 Source Files/Python/Question_4c.py`. Runs in the browser.*

Load `linprog`, SciPy's linear-programming solver. It always **minimises** $c^T x$ subject to `A_ub` $x \le$ `b_ub`, and by default every variable is $\ge 0$, so the non-negativity constraints need no rows of their own.

```python
from scipy.optimize import linprog
```

The same profits as in Q5b.

```python
# objective (row vector): linprog minimises, so to maximise we pass the negatives
c = [-12, -5, -8]
```

Only the machine time of X changes: 4.5 hours instead of 5.

```python
# linear inequalities 
# (no. rows in A = no. rows in b)
A_ub = [[4.5, 2, 3],   # machine A: 4.5x + 2y + 3z <= 100 hours
        [0, 0, 1]]     # material: z <= 20
b_ub = [100, 20]     
```

Solve and print. `results.success` is `True` when an optimum was found; `results.x` holds the optimal values of the variables and `results.fun` the minimised objective (`fval`).

```python
#  call the linprog function
results = linprog(c, A_ub, b_ub)

print(results.message)
# if results are a success then print out the optimal solution
if results.success:
    for k in range(len(results.x)):
        print(f"x[{k+1}] = {results.x[k]:.2f}")
    print(f"fval = {results.fun:.2f}")  
        
        
```

## What do you notice?

The question asks what you notice about this solution, and why it might not be the best plan in practice.

<details><summary>Answer</summary>

- **It is not a whole number.** You cannot make 22.22 units. Rounding down to $x = 22$ uses 99 hours and
  earns £264; the best whole-number plan earns £266, for example $x = 20$, $y = 2$, $z = 2$ (exactly
  100 hours). Rounding an LP answer does not give the best whole-number answer; Week 03 shows how to ask
  for whole numbers directly (integer programming).
- **It is not the only optimum.** X now earns $12/4.5 = 8/3 \approx £2.67$ per hour, exactly as much as Z,
  so the LP has many optimal solutions with profit £266.67 (for example $z = 20$, $x = 8.89$). A solver
  may report any of them.
