# Tutorial 2, Q3: the LP with SciPy's linprog

**The problem.** Tutorial Q1, now solved in Python:

maximise $5x + 17y$ subject to $2x + 3y \le 60$, $x + y \le 25$, $x, y \ge 0$.

**How to run it.** *Run → Run All Cells*, or press Shift + Enter in each cell while you read. Nothing to install: SciPy is built into the browser's Python (the first import takes a few seconds).

**Expected result.** `x[1] = 0.00`, `x[2] = 20.00`, `fval = -340.00`: $x = 0$, $y = 20$, maximum 340, the
vertex found graphically in Q1. `linprog` minimises, so the printed `fval` is **minus** the maximum (here
−340.00 means 340).

*Original script: `5 Source Files/Python/Question_3.py`. Runs in the browser.*

Load `linprog`, SciPy's linear-programming solver. It always **minimises** $c^T x$ subject to `A_ub` $x \le$ `b_ub`, and by default every variable is $\ge 0$, so the non-negativity constraints need no rows of their own.

```python
from scipy.optimize import linprog
```

`linprog` minimises, so to maximise $5x + 17y$ we minimise $-5x - 17y$.

```python
# objective (row vector): linprog minimises, so to maximise we pass the negatives
c = [-5, -17]
```

The constraints in the matrix form of Q2: one row per inequality, and row $k$ of `A_ub` goes with entry $k$ of `b_ub`.

```python
# linear inequalities 
# (no. rows in A = no. rows in b)
A_ub = [[2, 3],   # 2x + 3y <= 60
        [1, 1]]   # x + y <= 25
b_ub = [60, 25]     
```

Solve and print. `results.success` is `True` when an optimum was found; `results.x` holds the optimal values of the variables and `results.fun` the minimised objective (`fval`).

```python
#  call the linprog function
results = linprog(c, A_ub, b_ub)

print(results.message)
# if results are a success then print out the optimal solution
if results.success:
    for k in range(len(results.x)):
        print(f"x[{k+1}] = {results.x[k]:.2f}")
    print(f"fval = {results.fun:.2f}")  
        
```

## What do you notice?

Which constraints are binding at $(0, 20)$?

<details><summary>Answer</summary>

$2x + 3y = 60$ holds with equality, so it is binding; $x + y = 20 < 25$ leaves 5 units spare, so it is not
(`results.slack` is $(0, 5)$). The non-negativity constraint $x \ge 0$ is binding too: the optimum is the
vertex where $2x + 3y = 60$ meets the $y$-axis.
</details>
