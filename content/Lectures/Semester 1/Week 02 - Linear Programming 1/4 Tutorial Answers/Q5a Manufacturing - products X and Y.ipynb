# Tutorial 2, Q5a: manufacturing products X and Y

**The problem.** Machine A has 100 hours a week. Product X takes 5 hours and earns £12; product Y takes
2 hours and earns £5. With $x$ units of X and $y$ units of Y:

maximise $12x + 5y$ subject to $5x + 2y \le 100$, $x, y \ge 0$.

**How to run it.** *Run → Run All Cells*, or press Shift + Enter in each cell while you read. Nothing to install: SciPy is built into the browser's Python (the first import takes a few seconds).

**Expected result.** `x[1] = 0.00`, `x[2] = 50.00`, `fval = -250.00`: make 50 of Y and no X, profit £250.
`linprog` minimises, so the printed `fval` is **minus** the maximum profit (here −250.00 means £250.00).

Script name says 4a: the question was Q4 on an earlier sheet.

*Original script: `5 Source Files/Python/Question_4a.py`. Runs in the browser.*

Load `linprog`, SciPy's linear-programming solver. It always **minimises** $c^T x$ subject to `A_ub` $x \le$ `b_ub`, and by default every variable is $\ge 0$, so the non-negativity constraints need no rows of their own.

```python
from scipy.optimize import linprog
```

The profit per unit, negated: `linprog` minimises, so to maximise $12x + 5y$ we minimise $-12x - 5y$.

```python
# objective (row vector): linprog minimises, so to maximise we pass the negatives
c = [-12, -5]
```

One constraint, the machine hours. `A_ub` is a list of rows even when there is only one row.

```python
# linear inequalities 
# (no. rows in A = no. rows in b)
A_ub = [[5, 2]]   # machine A: 5x + 2y <= 100 hours
b_ub = [100]     
```

Solve and print. `results.success` is `True` when an optimum was found; `results.x` holds the optimal values of the variables and `results.fun` the minimised objective (`fval`).

```python
#  call the linprog function
results = linprog(c, A_ub, b_ub)

print(results.message)
# if results are a success then print out the optimal solution
if results.success:
    for k in range(len(results.x)):
        print(f"x[{k+1}] = {results.x[k]:.2f}")
    print(f"fval = {results.fun:.2f}")    
        
        
```

## What do you notice?

Y earns less per unit than X (£5 against £12), yet the answer makes only Y. Why?

<details><summary>Answer</summary>

Compare the profit per **machine hour**, the scarce resource: X earns $12/5 = £2.40$ an hour and Y
$5/2 = £2.50$ an hour. With a single constraint, every hour should go to the product that earns most per
hour: $100/2 = 50$ units of Y, $50 \times £5 = £250$.
</details>
