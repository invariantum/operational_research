# Tutorial 2, Q5b: adding product Z

**The problem.** Q5a plus a third product Z ($z$ units), which takes 3 hours and earns £8; there is
material for at most 20 units of Z a week:

maximise $12x + 5y + 8z$ subject to $5x + 2y + 3z \le 100$, $z \le 20$, $x, y, z \ge 0$.

**How to run it.** *Run → Run All Cells*, or press Shift + Enter in each cell while you read. Nothing to install: SciPy is built into the browser's Python (the first import takes a few seconds).

**Expected result.** `x[1] = 0.00`, `x[2] = 20.00`, `x[3] = 20.00`, `fval = -260.00`: 20 of Y and 20 of Z,
profit £260. `linprog` minimises, so the printed `fval` is **minus** the maximum profit (here −260.00 means
£260.00).

Script name says 4b: the question was Q4 on an earlier sheet.

*Original script: `5 Source Files/Python/Question_4b.py`. Runs in the browser.*

Load `linprog`, SciPy's linear-programming solver. It always **minimises** $c^T x$ subject to `A_ub` $x \le$ `b_ub`, and by default every variable is $\ge 0$, so the non-negativity constraints need no rows of their own.

```python
from scipy.optimize import linprog
```

Three variables now ($x$, $y$, $z$), so three negated profits.

```python
# objective (row vector): linprog minimises, so to maximise we pass the negatives
c = [-12, -5, -8]
```

Two constraints: machine hours and the material for Z. Each row has one entry per variable.

```python
# linear inequalities 
# (no. rows in A = no. rows in b)
A_ub = [[5, 2, 3],   # machine A: 5x + 2y + 3z <= 100 hours
        [0, 0, 1]]   # material: z <= 20
b_ub = [100, 20]     
```

Solve and print. `results.success` is `True` when an optimum was found; `results.x` holds the optimal values of the variables and `results.fun` the minimised objective (`fval`).

```python
#  call the linprog function
results = linprog(c, A_ub, b_ub)

print(results.message)
# if results are a success then print out the optimal solution
if results.success:
    for k in range(len(results.x)):
        print(f"x[{k+1}] = {results.x[k]:.2f}")
    print(f"fval = {results.fun:.2f}")    
        
        
```

## What do you notice?

Why 20 of Z and then 20 of Y, and still no X?

<details><summary>Answer</summary>

Per machine hour Z earns $8/3 \approx £2.67$, Y £2.50 and X £2.40. So Z gets the hours first, up to its
material limit: 20 units use 60 hours. The 40 hours left go to the next best product, Y: $40/2 = 20$ units.
Profit $20 \times 8 + 20 \times 5 = £260$, and both constraints are binding.
</details>
